# Projet Deep Learning — ChestScan AI : Classification de radiographies thoraciques

## Présentation du projet

**ChestScan AI** est une startup marocaine spécialisée dans les solutions d'intelligence artificielle appliquées à l'imagerie médicale. Elle souhaite développer un prototype capable d'analyser des radiographies thoraciques afin de distinguer les images classées comme **NORMAL** de celles associées à une **PNEUMONIA**.

L'objectif du projet est de mettre en œuvre plusieurs approches de Deep Learning pour la classification d'images, d'évaluer leurs performances et de comparer leurs résultats.

### Objectifs du projet

- Charger, explorer et préparer un ensemble de radiographies thoraciques.
- Concevoir et entraîner un réseau de neurones convolutif (CNN) à partir de zéro.
- Mettre en œuvre le Transfer Learning à l'aide d'un modèle préentraîné, MobileNetV2.
- Appliquer le Fine-tuning en réentraînant une partie des couches du modèle préentraîné.
- Évaluer et comparer les performances des différentes approches.
- Sauvegarder le modèle retenu et développer une interface simple permettant de tester le prototype sur de nouvelles images.
